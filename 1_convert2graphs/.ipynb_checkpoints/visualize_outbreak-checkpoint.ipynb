# Epidemic Outbreak Trajectory Visualization

This notebook visualizes the dynamics of epidemic simulations from BEAST2 outputs.

**Visualizations include:**
- Compartment dynamics over time (S, I, R)
- Individual location trajectories
- Total population dynamics
- Sample accumulation over time

**Input files:**
- `{i}_beast2.traj`: Trajectory data (population time series)
- `{i}_beast2.xml`: XML configuration (simulation parameters)

## Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Import our utility functions
from trajectory_utils import (
    load_trajectory_wide,
    load_parameters_from_xml,
    get_compartment_groups,
    get_sample_data,
    get_trajectory_summary
)

# Set plotting style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (14, 8)
plt.rcParams['font.size'] = 11
```

## Configuration

```python
# === CONFIGURE YOUR FILES HERE ===
TRAJ_FILE = '0_beast2.traj'  # Path to trajectory file
XML_FILE = '0_beast2.xml'    # Path to XML file
SAMPLE_ID = 0                 # Which simulation sample to visualize

# Check if files exist
if not Path(TRAJ_FILE).exists():
    print(f"⚠️  Trajectory file not found: {TRAJ_FILE}")
else:
    print(f"✓ Found trajectory file: {TRAJ_FILE}")

if not Path(XML_FILE).exists():
    print(f"⚠️  XML file not found: {XML_FILE}")
else:
    print(f"✓ Found XML file: {XML_FILE}")
```

## Load Data

```python
# Load trajectory data
print("Loading trajectory data...")
df_wide = load_trajectory_wide(TRAJ_FILE)
print(f"✓ Loaded {len(df_wide)} time points across {df_wide['Sample'].nunique()} samples")

# Load XML parameters
print("\nLoading simulation parameters...")
params = load_parameters_from_xml(XML_FILE)
print(f"✓ Number of locations: {params['num_locs']}")
print(f"✓ Number of simulations: {params['num_sims']}")
print(f"✓ Max simulation time: {params['max_time']:.2f}")

# Get trajectory summary
print("\nTrajectory summary:")
summary = get_trajectory_summary(TRAJ_FILE)
for key, value in summary.items():
    print(f"  {key}: {value}")
```

```python
# Extract data for specific sample
sample_data = get_sample_data(df_wide, SAMPLE_ID)
print(f"Extracted sample {SAMPLE_ID}: {len(sample_data)} time points")

# Get compartment groups
groups = get_compartment_groups(sample_data)
print(f"\nCompartments:")
print(f"  S compartments: {len(groups['S'])}")
print(f"  I compartments: {len(groups['I'])}")
print(f"  R compartment: {len(groups['R'])}")
print(f"  Sample compartment: {len(groups['sample'])}")
```

## Preview Data

```python
# Show first few rows
display(sample_data.head(10))
```

## Visualization 1: Total Population Dynamics (S, I, R)

```python
# Calculate totals across all locations
sample_data['Total_S'] = sample_data[groups['S']].sum(axis=1)
sample_data['Total_I'] = sample_data[groups['I']].sum(axis=1)
sample_data['Total_R'] = sample_data[groups['R']].sum(axis=1)
sample_data['Total_Sample'] = sample_data[groups['sample']].sum(axis=1)

# Plot
fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(sample_data['t'], sample_data['Total_S'], label='Susceptible (S)', 
        color='steelblue', linewidth=2)
ax.plot(sample_data['t'], sample_data['Total_I'], label='Infected (I)', 
        color='crimson', linewidth=2)
ax.plot(sample_data['t'], sample_data['Total_R'], label='Recovered (R)', 
        color='forestgreen', linewidth=2)
ax.plot(sample_data['t'], sample_data['Total_Sample'], label='Sampled', 
        color='orange', linewidth=2, linestyle='--')

ax.set_xlabel('Time', fontsize=12)
ax.set_ylabel('Population', fontsize=12)
ax.set_title(f'Total Population Dynamics - Sample {SAMPLE_ID}', fontsize=14, fontweight='bold')
ax.legend(loc='best', fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Print final statistics
print(f"\nFinal Statistics (t = {sample_data['t'].max():.2f}):")
print(f"  Susceptible: {sample_data['Total_S'].iloc[-1]:.0f}")
print(f"  Infected: {sample_data['Total_I'].iloc[-1]:.0f}")
print(f"  Recovered: {sample_data['Total_R'].iloc[-1]:.0f}")
print(f"  Sampled: {sample_data['Total_Sample'].iloc[-1]:.0f}")
```

## Visualization 2: Infected Population by Location

```python
fig, ax = plt.subplots(figsize=(14, 7))

# Plot each infected compartment
for i_comp in groups['I']:
    ax.plot(sample_data['t'], sample_data[i_comp], label=i_comp, linewidth=1.5, alpha=0.8)

ax.set_xlabel('Time', fontsize=12)
ax.set_ylabel('Number of Infected Individuals', fontsize=12)
ax.set_title(f'Infected Population by Location - Sample {SAMPLE_ID}', fontsize=14, fontweight='bold')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=9)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Visualization 3: Susceptible Population by Location

```python
fig, ax = plt.subplots(figsize=(14, 7))

# Plot each susceptible compartment
for s_comp in groups['S']:
    ax.plot(sample_data['t'], sample_data[s_comp], label=s_comp, linewidth=1.5, alpha=0.8)

ax.set_xlabel('Time', fontsize=12)
ax.set_ylabel('Number of Susceptible Individuals', fontsize=12)
ax.set_title(f'Susceptible Population by Location - Sample {SAMPLE_ID}', fontsize=14, fontweight='bold')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=9)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Visualization 4: Heatmap of Infected Population Over Time

```python
# Prepare data for heatmap
infected_data = sample_data[['t'] + groups['I']].set_index('t')

# Rename columns to just location numbers for cleaner display
infected_data.columns = [col.replace('I[', '').replace(']', '') for col in infected_data.columns]

fig, ax = plt.subplots(figsize=(14, 8))

# Create heatmap
sns.heatmap(infected_data.T, cmap='YlOrRd', cbar_kws={'label': 'Number of Infected'}, ax=ax)

ax.set_xlabel('Time Index', fontsize=12)
ax.set_ylabel('Location', fontsize=12)
ax.set_title(f'Infected Population Heatmap - Sample {SAMPLE_ID}', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

## Visualization 5: Stacked Area Plot - Compartment Distribution

```python
fig, ax = plt.subplots(figsize=(14, 7))

# Create stacked area plot
ax.fill_between(sample_data['t'], 0, sample_data['Total_S'], 
                 label='Susceptible', color='steelblue', alpha=0.7)
ax.fill_between(sample_data['t'], sample_data['Total_S'], 
                 sample_data['Total_S'] + sample_data['Total_I'], 
                 label='Infected', color='crimson', alpha=0.7)
ax.fill_between(sample_data['t'], 
                 sample_data['Total_S'] + sample_data['Total_I'],
                 sample_data['Total_S'] + sample_data['Total_I'] + sample_data['Total_R'],
                 label='Recovered', color='forestgreen', alpha=0.7)

ax.set_xlabel('Time', fontsize=12)
ax.set_ylabel('Population', fontsize=12)
ax.set_title(f'Compartment Distribution Over Time - Sample {SAMPLE_ID}', fontsize=14, fontweight='bold')
ax.legend(loc='upper right', fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Visualization 6: Peak Infection Analysis

```python
# Find peak infection for each location
peak_data = []
for i_comp in groups['I']:
    loc_num = i_comp.replace('I[', '').replace(']', '')
    peak_idx = sample_data[i_comp].idxmax()
    peak_time = sample_data.loc[peak_idx, 't']
    peak_value = sample_data.loc[peak_idx, i_comp]
    peak_data.append({
        'Location': loc_num,
        'Peak_Time': peak_time,
        'Peak_Infected': peak_value
    })

peak_df = pd.DataFrame(peak_data)

# Plot peak infections
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Peak infection by location
ax1.bar(peak_df['Location'], peak_df['Peak_Infected'], color='crimson', alpha=0.7)
ax1.set_xlabel('Location', fontsize=12)
ax1.set_ylabel('Peak Number of Infected', fontsize=12)
ax1.set_title('Peak Infection by Location', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3, axis='y')

# Peak timing by location
ax2.bar(peak_df['Location'], peak_df['Peak_Time'], color='steelblue', alpha=0.7)
ax2.set_xlabel('Location', fontsize=12)
ax2.set_ylabel('Time of Peak Infection', fontsize=12)
ax2.set_title('Timing of Peak Infection by Location', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# Display peak data table
print("\nPeak Infection Summary:")
display(peak_df)
```

## Visualization 7: Compare Multiple Samples (Optional)

```python
# Compare total infected across multiple samples
num_samples_to_compare = min(5, df_wide['Sample'].nunique())  # Compare up to 5 samples

fig, ax = plt.subplots(figsize=(14, 7))

for sample_id in range(num_samples_to_compare):
    sample = get_sample_data(df_wide, sample_id)
    total_infected = sample[groups['I']].sum(axis=1)
    ax.plot(sample['t'], total_infected, label=f'Sample {sample_id}', linewidth=2, alpha=0.7)

ax.set_xlabel('Time', fontsize=12)
ax.set_ylabel('Total Infected', fontsize=12)
ax.set_title('Comparison of Infected Population Across Samples', fontsize=14, fontweight='bold')
ax.legend(loc='best', fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Summary Statistics

```python
print("=" * 60)
print(f"OUTBREAK SUMMARY - Sample {SAMPLE_ID}")
print("=" * 60)

print(f"\nSimulation Duration: {sample_data['t'].max():.2f} time units")

print(f"\nInitial State (t=0):")
print(f"  Total Susceptible: {sample_data['Total_S'].iloc[0]:.0f}")
print(f"  Total Infected: {sample_data['Total_I'].iloc[0]:.0f}")

print(f"\nFinal State:")
print(f"  Total Susceptible: {sample_data['Total_S'].iloc[-1]:.0f}")
print(f"  Total Infected: {sample_data['Total_I'].iloc[-1]:.0f}")
print(f"  Total Recovered: {sample_data['Total_R'].iloc[-1]:.0f}")
print(f"  Total Sampled: {sample_data['Total_Sample'].iloc[-1]:.0f}")

print(f"\nPeak Infection:")
peak_total_idx = sample_data['Total_I'].idxmax()
print(f"  Time: {sample_data.loc[peak_total_idx, 't']:.2f}")
print(f"  Number: {sample_data.loc[peak_total_idx, 'Total_I']:.0f}")

print(f"\nAttack Rate: {((sample_data['Total_R'].iloc[-1] + sample_data['Total_Sample'].iloc[-1]) / sample_data['Total_S'].iloc[0] * 100):.1f}%")

print("\n" + "=" * 60)
```
