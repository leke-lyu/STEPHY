# DTW+S Edge Attention Pipeline

```
Tip times → KDE → curve → SSR → DTW → (distance, lag, lag_std)
```

Reference: [DTW+S](https://arxiv.org/abs/2309.03579)

````python
import re
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
from scipy.stats import gaussian_kde
````

## Step 1: Tip Times → KDE → Curve

````python
def parse_tree(tree_file, tree_idx=0):
    """Parse a single tree from a BEAST2 .trees file."""
    with open(tree_file, 'r') as f:
        content = f.read()
    
    tree_pattern = r'tree STATE_\d+ = (.+?)(?=\ntree |\nEnd;|$)'
    matches = re.findall(tree_pattern, content, re.DOTALL)
    return matches[tree_idx].strip()


def extract_tip_times(tree_string):
    """Extract tip times grouped by location from a tree string."""
    pattern = r'\d+\[&type="I\{(\d+)\}",samp="sample",time=([\d.]+)\]'
    matches = re.findall(pattern, tree_string)
    
    tips_by_location = defaultdict(list)
    for loc, time in matches:
        tips_by_location[int(loc)].append(float(time))
    
    return dict(tips_by_location)


def compute_epidemic_curves(tips_by_location, n_points=200):
    """Convert tip times to epidemic curves via KDE (Scott's rule)."""
    all_times = [t for times in tips_by_location.values() for t in times]
    t_min, t_max = min(all_times), max(all_times)
    t_grid = np.linspace(t_min, t_max, n_points)
    
    curves = {}
    for loc, times in tips_by_location.items():
        if len(times) >= 2:
            kde = gaussian_kde(times)
            curves[loc] = kde(t_grid)
        else:
            curves[loc] = np.zeros(n_points)
    
    return t_grid, curves
````

````python
# Load tree and compute curves
tree_file = '/Users/lukelyu/Desktop/epidata/20_5/0_beast2.trees'
tree_string = parse_tree(tree_file, tree_idx=0)
tips_by_location = extract_tip_times(tree_string)
t_grid, curves = compute_epidemic_curves(tips_by_location)
````

````python
# Plot epidemic curves
fig, axes = plt.subplots(3, 4, figsize=(14, 8))

for idx, loc in enumerate(sorted(curves.keys())):
    ax = axes.flatten()[idx]
    ax.plot(t_grid, curves[loc], 'b-', lw=1.5)
    ax.fill_between(t_grid, curves[loc], alpha=0.3)
    ax.set_title(f'Loc {loc} (n={len(tips_by_location[loc])})')
    ax.set_xlabel('Time')
    ax.set_ylabel('Density')

plt.tight_layout()
plt.show()
````

````python
# Overlay all curves
plt.figure(figsize=(10, 5))
colors = plt.cm.tab20(np.linspace(0, 1, len(curves)))

for idx, loc in enumerate(sorted(curves.keys())):
    plt.plot(t_grid, curves[loc], color=colors[idx], lw=1.5, label=f'Loc {loc}', alpha=0.7)

plt.xlabel('Time')
plt.ylabel('Density')
plt.title('Epidemic Curves - All Locations')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
````
