# Root vs MRCA vs Earliest Tip Analysis

Compare location labels of:
- **Root**: `tree.seed_node` (origin, earliest time)
- **MRCA**: `tree.mrca()` (most recent common ancestor of all tips)
- **Earliest Tip**: tip with smallest time

```python
import re, glob, dendropy, pandas as pd, warnings
warnings.filterwarnings('ignore')

DATA_FOLDER = "/Users/lukelyu/Desktop/epidata/500_1_0.001"
```

```python
def load_tree(filepath):
    """Load BEAST2 tree and convert to dendropy format."""
    with open(filepath) as f:
        content = f.read()
    tree_str = re.search(r'tree\s+\w+\s*=\s*(.+)', content).group(1)
    # Convert BEAST2 annotations: [&type="I{8}",time=X] -> [&location=8,time=X]
    converted = re.sub(r'\[&[^\]]+\]', lambda m: 
        f'[&location={re.search(r"I\{(\d+)\}", m.group(0)).group(1)},'
        f'time={re.search(r"time=([\d.eE+-]+)", m.group(0)).group(1)}]', tree_str)
    tree = dendropy.Tree.get(data=converted, schema="newick", extract_comment_metadata=True)
    tree.is_rooted = True
    return tree

def analyze_tree(filepath):
    """Extract root, MRCA, earliest tip info from a tree."""
    tree = load_tree(filepath)
    tips = list(tree.leaf_node_iter())
    
    root = tree.seed_node
    mrca = tree.mrca(taxon_labels=[t.taxon.label for t in tips])
    earliest = min(tips, key=lambda t: float(t.annotations.get_value('time')))
    
    get_loc = lambda n: int(n.annotations.get_value('location'))
    get_time = lambda n: float(n.annotations.get_value('time'))
    
    return {
        'file': filepath.split('/')[-1].replace('_beast2.trees', ''),
        'root_loc': get_loc(root), 'root_time': get_time(root),
        'mrca_loc': get_loc(mrca), 'mrca_time': get_time(mrca),
        'etip_loc': get_loc(earliest), 'etip_time': get_time(earliest)
    }
```

```python
# Analyze all trees
files = sorted(glob.glob(f"{DATA_FOLDER}/*_beast2.trees"))
results = [analyze_tree(f) for i, f in enumerate(files) if not (i % 100 == 99 and print(f"{i+1}/{len(files)}..."))]
df = pd.DataFrame(results)
print(f"Analyzed {len(df)} trees")
```

```python
# Summary
n = len(df)
df['root_eq_mrca'] = df['root_loc'] == df['mrca_loc']
df['mrca_eq_etip'] = df['mrca_loc'] == df['etip_loc']
df['root_eq_etip'] = df['root_loc'] == df['etip_loc']

print(f"Root == MRCA:         {df['root_eq_mrca'].sum():3d}/{n} ({100*df['root_eq_mrca'].mean():.1f}%)")
print(f"MRCA == Earliest Tip: {df['mrca_eq_etip'].sum():3d}/{n} ({100*df['mrca_eq_etip'].mean():.1f}%)")
print(f"Root == Earliest Tip: {df['root_eq_etip'].sum():3d}/{n} ({100*df['root_eq_etip'].mean():.1f}%)")
print(f"All Three Match:      {(df['root_eq_mrca'] & df['mrca_eq_etip']).sum():3d}/{n}")
```
