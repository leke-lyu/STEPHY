# Earliest Tip vs Ancestral State

For each simulation in a dataset, check whether the **earliest sampled tip**
(the tip with the smallest forward time in the BEAST2 tree) belongs to the
same location as the **ancestral state** (the outbreak origin labeled in `*_nf.csv`).

Reports the match rate per dataset.

```python
import re
import csv
from pathlib import Path

# Tip annotation: [&type="I{LOC}",samp="sample",time=T]
TIP_RE = re.compile(r'\[&type="I\{(\d+)\}",samp="sample",time=([\d.]+)\]')
TREE_RE = re.compile(r'tree STATE_\d+ = (.+?)\n')


def earliest_tip_location(trees_path):
    """Location (int) of the earliest sampled tip in the first tree."""
    tree_str = TREE_RE.search(Path(trees_path).read_text()).group(1)
    loc, _ = min(TIP_RE.findall(tree_str), key=lambda x: float(x[1]))
    return int(loc)


def ancestral_state_location(nf_path):
    """Location (int) where Ancestral_State == 1."""
    with open(nf_path) as f:
        for row in csv.DictReader(f):
            if row['Ancestral_State'] == '1':
                return int(row['Location'])
    return None


def match_rate(dataset_dir):
    """Fraction of simulations where earliest tip == ancestral state."""
    dataset_dir = Path(dataset_dir)
    matches, total = 0, 0
    for tf in sorted(dataset_dir.glob('*_beast2.trees')):
        nf = dataset_dir / f'{tf.name.replace("_beast2.trees", "")}_nf.csv'
        if not nf.exists():
            continue
        matches += earliest_tip_location(tf) == ancestral_state_location(nf)
        total += 1
    return matches, total
```

```python
datasets = [
    '/Users/lukelyu/Desktop/epidata/1000_1_0.0006_0.0007/',
    '/Users/lukelyu/Desktop/epidata/1000_1_0.0006_0.0015/',
    '/Users/lukelyu/Desktop/epidata/1000_1_0.0006_0.0023/'
    # add more dataset paths here
]

for ds in datasets:
    matches, total = match_rate(ds)
    print(f'{Path(ds).name:40s}  {matches}/{total}  ({matches/total:.2%})')
```
