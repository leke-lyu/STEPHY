```python
import re
import numpy as np
from pathlib import Path
from collections import defaultdict

DATASET = '/Users/lukelyu/Desktop/epidata/batch_0/'
TOP_PCT = 5  # percentage of largest trees to discard

# --- Parse all trees -----------------------------------------------------------
TREE_RE = re.compile(r'tree STATE_\d+ = (.+?)(?=\ntree |\nEnd;|$)', re.DOTALL)
TIP_RE  = re.compile(r'\d+\[&type="I\{(\d+)\}",samp="sample"')

tree_stats = []  # (tree_id, total_tips, {loc: count})
for tf in sorted(Path(DATASET).glob('*_beast2.trees')):
    prefix = tf.stem.replace('_beast2', '')
    for idx, tree_str in enumerate(TREE_RE.findall(tf.read_text())):
        loc_counts = defaultdict(int)
        for loc in TIP_RE.findall(tree_str):
            loc_counts[int(loc)] += 1
        tree_stats.append((f'{prefix}_{idx}', sum(loc_counts.values()), dict(loc_counts)))

all_locations = sorted({loc for _, _, lc in tree_stats for loc in lc})
num_locations = len(all_locations)

# --- Helper: print tree_width and subtree_width stats -------------------------
def print_stats(stats):
    sizes = np.array([t[1] for t in stats])
    min_t = min(stats, key=lambda t: t[1])
    max_t = max(stats, key=lambda t: t[1])

    subs = [(tid, loc, cnt) for tid, _, lc in stats for loc, cnt in lc.items()]
    sub_sizes = np.array([x[2] for x in subs])
    min_s = min(subs, key=lambda x: x[2])
    max_s = max(subs, key=lambda x: x[2])

    print(f'  tree_width     — Range: [{min_t[1]}, {max_t[1]}],  Mean: {sizes.mean():.1f}')
    print(f'    Min: {min_t[0]} ({min_t[1]} tips)')
    print(f'    Max: {max_t[0]} ({max_t[1]} tips)')
    print(f'  subtree_width  — Range: [{min_s[2]}, {max_s[2]}],  Mean: {sub_sizes.mean():.1f}')
    print(f'    Min: tree={min_s[0]}, loc={min_s[1]} ({min_s[2]} tips)')
    print(f'    Max: tree={max_s[0]}, loc={max_s[1]} ({max_s[2]} tips)')
    return max_s[2]

# --- Report --------------------------------------------------------------------
print(f'Dataset:   {DATASET}')
print(f'Locations: {all_locations}')

print(f'\n=== All trees ({len(tree_stats)} trees) ===')
print_stats(tree_stats)

cutoff = int(np.percentile([t[1] for t in tree_stats], 100 - TOP_PCT))
filtered = [t for t in tree_stats if t[1] <= cutoff]

print(f'\n=== Discarding top {TOP_PCT}% ({len(filtered)} trees <= {cutoff} tips) ===')
sw = print_stats(filtered)

print(f'\n--num_locations {num_locations} --subtree_width {sw} --tree_size_threshold {cutoff}')
```
