# Patristic Distance Explorer

This notebook analyzes patristic distance distributions from phylogenetic graph data.
For each edge in a specified graph, we compute and visualize summary statistics of the distance arrays.

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style for better visualizations
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 8)
```

## Configuration

Specify the path to your edge CSV file and the graph_id you want to analyze.

```python
# Path to the edge CSV file
edge_file_path = "/Users/lukelyu/Desktop/epidata/100_10_new/10_edge.csv"

# Target graph to analyze (change this to explore different graphs)
target_graph_id = "10_1"
```

## Load and Filter Data

Load the CSV file and filter for the target graph.

```python
# Load the edge CSV file
df = pd.read_csv(edge_file_path)

print(f"Total edges in file: {len(df)}")
print(f"Unique graph IDs: {df['graph_id'].nunique()}")
print(f"Available graph IDs: {sorted(df['graph_id'].unique())}")
```

```python
# Filter edges for the target graph
df_filtered = df[df['graph_id'] == target_graph_id].copy()

print(f"\nFiltered to graph '{target_graph_id}': {len(df_filtered)} edges")

if len(df_filtered) == 0:
    print(f"\nWarning: No edges found for graph_id '{target_graph_id}'")
    print(f"Please choose from: {sorted(df['graph_id'].unique())}")
```

```python
def parse_patristic_distances(distance_string):
    """
    Parse patristic distances from array string format [value1 value2 ...].
    
    Parameters:
    -----------
    distance_string : str
        String representation of distance array
        
    Returns:
    --------
    np.array or None
        Array of distances, or None if invalid/empty
    """
    distance_string = str(distance_string).strip()
    
    # Check if empty or NaN
    if not distance_string or distance_string.lower() == 'nan' or distance_string == '':
        return None
    
    # Remove brackets from array format
    if distance_string.startswith('['):
        distance_string = distance_string[1:]
    if distance_string.endswith(']'):
        distance_string = distance_string[:-1]
    
    # Parse values
    try:
        values = [float(x.strip()) for x in distance_string.split() if x.strip()]
        if len(values) == 0:
            return None
        return np.array(values)
    except:
        return None

# Parse patristic distances from string format to arrays
df_filtered['distances_array'] = df_filtered['patristic_distance'].apply(parse_patristic_distances)

# Filter out edges with invalid/empty patristic distances
original_count = len(df_filtered)
df_filtered = df_filtered[df_filtered['distances_array'].notna()].copy()
dropped_count = original_count - len(df_filtered)

# Report filtering results
print(f"Edges with valid patristic distances: {len(df_filtered)}")
if dropped_count > 0:
    print(f"Dropped {dropped_count} edges with missing/invalid patristic_distance values")
```

## Calculate Statistics

Compute summary statistics (min, 25%, median, 75%, max) for each edge's distance distribution.

```python
def calculate_statistics(distances_array):
    """
    Calculate distribution statistics for a distance array.
    
    Parameters:
    -----------
    distances_array : np.array
        Array of patristic distances
        
    Returns:
    --------
    dict
        Dictionary containing min, 25%, median, 75%, max
    """
    return {
        'min': np.min(distances_array),
        'q25': np.percentile(distances_array, 25),
        'median': np.median(distances_array),
        'q75': np.percentile(distances_array, 75),
        'max': np.max(distances_array)
    }

# Calculate statistics for each edge
stats_list = []
for idx, row in df_filtered.iterrows():
    stats = calculate_statistics(row['distances_array'])
    stats['edge_label'] = f"{row['src']}-{row['dst']}"
    stats['src'] = row['src']
    stats['dst'] = row['dst']
    stats['distance_count'] = len(row['distances_array'])
    stats_list.append(stats)

# Create statistics dataframe
stats_df = pd.DataFrame(stats_list)

# Display summary
print(f"Statistics computed for {len(stats_df)} edges\n")
print("Sample statistics (first 10 edges):")
print(stats_df.head(10).to_string())

print(f"\n{'='*70}")
print("Overall Summary Across All Edges:")
print(f"{'='*70}")
print(f"  Minimum distance value: {stats_df['min'].min():.4f}")
print(f"  Maximum distance value: {stats_df['max'].max():.4f}")
print(f"  Average median distance: {stats_df['median'].mean():.4f}")
print(f"  Total distance values: {stats_df['distance_count'].sum()}")
```

## Visualization

Box plot showing the distribution of patristic distances for each edge.

```python
# Sort edges by minimum distance value for better visualization
stats_df_sorted = stats_df.sort_values('min').reset_index(drop=True)

# Create figure and axis
fig, ax = plt.subplots(figsize=(16, 8))

# Prepare box plot data
positions = range(len(stats_df_sorted))
bp_dicts = []

for idx, row in stats_df_sorted.iterrows():
    bp_dict = {
        'label': row['edge_label'],
        'whislo': row['min'],      # Minimum
        'q1': row['q25'],          # 25th percentile
        'med': row['median'],      # Median
        'q3': row['q75'],          # 75th percentile
        'whishi': row['max'],      # Maximum
        'fliers': []               # No outliers
    }
    bp_dicts.append(bp_dict)

# Create box plot
bp = ax.bxp(bp_dicts, positions=positions, widths=0.6, 
            patch_artist=True, showfliers=False)

# Customize appearance
for patch in bp['boxes']:
    patch.set_facecolor('lightblue')
    patch.set_alpha(0.7)

# Set labels and title
ax.set_xlabel('Edge (sorted by minimum distance)', fontsize=12, fontweight='bold')
ax.set_ylabel('Patristic Distance', fontsize=12, fontweight='bold')
ax.set_title(f'Patristic Distance Distribution for Graph {target_graph_id}', 
             fontsize=14, fontweight='bold')

# Set x-axis labels
ax.set_xticks(positions)
ax.set_xticklabels(stats_df_sorted['edge_label'], rotation=45, ha='right')
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
