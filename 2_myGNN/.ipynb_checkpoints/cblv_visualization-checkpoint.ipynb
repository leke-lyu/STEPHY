# Location-Specific CBLV Extraction: Visual Demonstration

This notebook demonstrates how CBLV (Compact Bijective Ladderized Vector) encoding works for location-specific virtual subtrees.

**Key concepts:**
1. **Virtual subtree**: For each location, extract only tips from that location
2. **Branch collapsing**: Internal nodes with only 1 relevant child are collapsed
3. **Ladderization**: Children sorted by max root distance (descending) for deterministic encoding
4. **Strict bifurcation**: Virtual subtree is always binary (n tips → n-1 internal nodes)

```python
# Cell 1: Setup and Tree Simulation

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from collections import defaultdict
import random

random.seed(42)
np.random.seed(42)

# Simple tree node class for simulation
class TreeNode:
    _id_counter = 0
    
    def __init__(self, is_leaf=False, location=None, root_distance=0):
        self.id = TreeNode._id_counter
        TreeNode._id_counter += 1
        self.is_leaf = is_leaf
        self.location = location  # Only for leaves: 'A', 'B', 'C'
        self.root_distance = root_distance
        self.children = []
        self.parent = None
        self.edge_length = 0
        # For CBLV encoding
        self.loc_counts = {}
        self.loc_max_dist = {}
        self.name = None
    
    def add_child(self, child):
        self.children.append(child)
        child.parent = self

def simulate_tree(n_tips=30, locations=['A', 'B', 'C']):
    """
    Simulate a random bifurcating tree with tips assigned to locations.
    Uses a simple coalescent-like process.
    """
    TreeNode._id_counter = 0
    
    # Create tips with random locations
    tips = []
    for i in range(n_tips):
        loc = random.choice(locations)
        tip = TreeNode(is_leaf=True, location=loc)
        tip.name = f"T{i}({loc})"
        tips.append(tip)
    
    # Build tree by random coalescence
    nodes = tips.copy()
    current_time = 0
    internal_count = 0
    
    while len(nodes) > 1:
        # Pick two random nodes to coalesce
        idx1, idx2 = random.sample(range(len(nodes)), 2)
        if idx1 > idx2:
            idx1, idx2 = idx2, idx1
        
        node1 = nodes.pop(idx2)
        node2 = nodes.pop(idx1)
        
        # Create parent node
        current_time += random.uniform(0.5, 2.0)
        parent = TreeNode(is_leaf=False, root_distance=current_time)
        parent.name = f"N{internal_count}"
        internal_count += 1
        
        # Set edge lengths
        node1.edge_length = current_time - node1.root_distance
        node2.edge_length = current_time - node2.root_distance
        
        parent.add_child(node1)
        parent.add_child(node2)
        
        nodes.append(parent)
    
    root = nodes[0]
    
    # Update root distances (from root, not to root)
    tree_height = root.root_distance
    def update_distances(node, dist_from_root=0):
        node.root_distance = dist_from_root
        for child in node.children:
            update_distances(child, dist_from_root + child.edge_length)
    update_distances(root, 0)
    
    return root, tips, tree_height

# Simulate tree
root, tips, tree_height = simulate_tree(n_tips=30, locations=['A', 'B', 'C'])

# Count tips per location
loc_counts = defaultdict(int)
for tip in tips:
    loc_counts[tip.location] += 1

print(f"Tree simulated with {len(tips)} tips")
print(f"Tree height: {max(t.root_distance for t in tips):.2f}")
print(f"Tips per location: {dict(loc_counts)}")
```

```python
# Cell 2: Tree Visualization Functions

def get_all_nodes(root):
    """Get all nodes in the tree."""
    nodes = []
    def traverse(node):
        nodes.append(node)
        for child in node.children:
            traverse(child)
    traverse(root)
    return nodes

def compute_y_positions(root):
    """Compute y positions for tree visualization (leaves evenly spaced)."""
    y_pos = {}
    leaf_count = [0]
    
    def assign_y(node):
        if node.is_leaf:
            y_pos[node.id] = leaf_count[0]
            leaf_count[0] += 1
        else:
            for child in node.children:
                assign_y(child)
            # Internal node y = mean of children y
            child_ys = [y_pos[c.id] for c in node.children]
            y_pos[node.id] = np.mean(child_ys)
    
    assign_y(root)
    return y_pos

def plot_tree(root, ax, title="Phylogenetic Tree", highlight_loc=None, 
              show_labels=True, color_map=None):
    """
    Plot tree with tips colored by location.
    If highlight_loc is specified, gray out other locations.
    """
    if color_map is None:
        color_map = {'A': '#e41a1c', 'B': '#377eb8', 'C': '#4daf4a'}
    
    y_pos = compute_y_positions(root)
    all_nodes = get_all_nodes(root)
    
    # Draw edges
    for node in all_nodes:
        if node.parent is not None:
            x1, x2 = node.parent.root_distance, node.root_distance
            y1, y2 = y_pos[node.parent.id], y_pos[node.id]
            
            # Determine edge color
            edge_color = 'gray'
            edge_alpha = 0.3
            linewidth = 1
            
            if highlight_loc is None:
                edge_color = 'black'
                edge_alpha = 0.7
            elif node.loc_counts.get(highlight_loc, 0) > 0:
                edge_color = color_map.get(highlight_loc, 'black')
                edge_alpha = 1.0
                linewidth = 2
            
            # Draw L-shaped connection
            ax.plot([x1, x1], [y1, y2], color=edge_color, alpha=edge_alpha, lw=linewidth)
            ax.plot([x1, x2], [y2, y2], color=edge_color, alpha=edge_alpha, lw=linewidth)
    
    # Draw nodes
    for node in all_nodes:
        x, y = node.root_distance, y_pos[node.id]
        
        if node.is_leaf:
            color = color_map.get(node.location, 'gray')
            alpha = 1.0
            size = 80
            
            if highlight_loc is not None and node.location != highlight_loc:
                color = 'lightgray'
                alpha = 0.5
                size = 40
            
            ax.scatter(x, y, c=color, s=size, zorder=5, alpha=alpha, edgecolors='black', linewidths=0.5)
            
            if show_labels and (highlight_loc is None or node.location == highlight_loc):
                ax.annotate(node.name, (x, y), xytext=(5, 0), textcoords='offset points',
                           fontsize=7, va='center')
    
    ax.set_xlabel('Root Distance', fontsize=10)
    ax.set_ylabel('Tips', fontsize=10)
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    
    return y_pos

print("Visualization functions defined")
```

```python
# Cell 3: CBLV Encoder (adapted from cblv_node_classification.ipynb)

def preprocess_tree(root):
    """
    Compute loc_counts and loc_max_dist for each node (bottom-up).
    This is equivalent to _preprocess() in the original code.
    """
    def postorder(node):
        if node.is_leaf:
            if node.location:
                node.loc_counts = {node.location: 1}
                node.loc_max_dist = {node.location: node.root_distance}
            else:
                node.loc_counts = {}
                node.loc_max_dist = {}
        else:
            node.loc_counts = {}
            node.loc_max_dist = {}
            for child in node.children:
                postorder(child)
                for loc, count in child.loc_counts.items():
                    node.loc_counts[loc] = node.loc_counts.get(loc, 0) + count
                for loc, dist in child.loc_max_dist.items():
                    node.loc_max_dist[loc] = max(node.loc_max_dist.get(loc, 0), dist)
    
    postorder(root)

def find_mrca(root, target_loc):
    """
    Find MRCA of all tips with target location.
    """
    total = root.loc_counts.get(target_loc, 0)
    if total < 2:
        return None
    
    mrca = root
    while True:
        children_with_all = [c for c in mrca.children 
                            if c.loc_counts.get(target_loc, 0) == total]
        if len(children_with_all) == 1:
            mrca = children_with_all[0]
        else:
            break
    return mrca

def is_branching_point(node, target_loc):
    """
    Check if node is a branching point for target_loc.
    True if >= 2 children have tips from target_loc.
    """
    if node.is_leaf:
        return False
    return sum(1 for c in node.children if c.loc_counts.get(target_loc, 0) > 0) >= 2

def find_parent_branching_point(node, target_loc, mrca):
    """Find nearest ancestor that is a branching point."""
    current = node.parent
    while current:
        if current == mrca or is_branching_point(current, target_loc):
            return current
        current = current.parent
    return mrca

def compute_accumulated_edge(node, parent_branch):
    """Sum edge lengths from node up to parent branching point."""
    total = 0
    current = node
    while current != parent_branch and current:
        total += current.edge_length
        current = current.parent
    return total

def virtual_inorder(node, target_loc, last_branch_dist, mrca, trace=None):
    """
    Generator for virtual in-order traversal of location subtree.
    Includes trace for visualization.
    """
    if trace is None:
        trace = []
    
    if node.is_leaf:
        if node.location == target_loc:
            parent_branch = find_parent_branching_point(node, target_loc, mrca)
            accum_edge = compute_accumulated_edge(node, parent_branch)
            trace.append(('leaf', node, node.root_distance - last_branch_dist, accum_edge))
            yield ('leaf', node.root_distance - last_branch_dist, accum_edge, node)
    
    elif is_branching_point(node, target_loc):
        # LADDERIZATION: Sort children by loc_max_dist descending
        children = sorted(
            [c for c in node.children if c.loc_counts.get(target_loc, 0) > 0],
            key=lambda c: c.loc_max_dist.get(target_loc, 0),
            reverse=True
        )
        
        # Visit left (deepest) subtree first
        yield from virtual_inorder(children[0], target_loc, last_branch_dist, mrca, trace)
        
        # Yield this internal node
        if node == mrca:
            accum = node.edge_length
        else:
            parent_branch = find_parent_branching_point(node, target_loc, mrca)
            accum = compute_accumulated_edge(node, parent_branch)
        
        trace.append(('internal', node, node.root_distance, accum))
        yield ('internal', node.root_distance, accum, node)
        
        # Visit remaining children
        for child in children[1:]:
            yield from virtual_inorder(child, target_loc, node.root_distance, mrca, trace)
    
    else:
        # COLLAPSE: Only 1 relevant child, pass through
        relevant = [c for c in node.children if c.loc_counts.get(target_loc, 0) > 0]
        if relevant:
            trace.append(('collapse', node))
            yield from virtual_inorder(relevant[0], target_loc, last_branch_dist, mrca, trace)

def encode_cblv(root, target_loc, tree_height):
    """
    Encode CBLV for a target location.
    Returns: (cblv_matrix, trace, mrca)
    """
    mrca = find_mrca(root, target_loc)
    if mrca is None:
        n_tips = root.loc_counts.get(target_loc, 0)
        return np.zeros((max(n_tips, 1), 4)), [], None
    
    n_tips = mrca.loc_counts.get(target_loc, 0)
    heights = np.zeros((n_tips, 4))
    stem = mrca.root_distance
    
    trace = []
    idx = 0
    
    for event in virtual_inorder(mrca, target_loc, mrca.root_distance, mrca, trace):
        if idx >= n_tips:
            break
        
        event_type, val1, val2, node = event
        
        if event_type == 'leaf':
            heights[idx, 0] = val1 + (stem if idx == 0 else 0)
            heights[idx, 2] = val2
        else:  # internal
            if idx + 1 < n_tips:
                heights[idx + 1, 1] = val1
                heights[idx + 1, 3] = val2
            idx += 1
    
    # Rescale by tree height
    heights = heights / tree_height
    
    return heights, trace, mrca

# Preprocess tree
preprocess_tree(root)
print("CBLV encoder ready")
print(f"\nRoot loc_counts: {root.loc_counts}")
print(f"Root loc_max_dist: {dict((k, f'{v:.2f}') for k,v in root.loc_max_dist.items())}")
```

```python
# Cell 4: Visualize Original Tree

fig, ax = plt.subplots(figsize=(14, 10))

y_pos = plot_tree(root, ax, title="Original Tree (30 tips, 3 locations)", show_labels=True)

# Add legend
color_map = {'A': '#e41a1c', 'B': '#377eb8', 'C': '#4daf4a'}
legend_patches = [mpatches.Patch(color=color_map[loc], label=f'Location {loc} ({loc_counts[loc]} tips)') 
                  for loc in ['A', 'B', 'C']]
ax.legend(handles=legend_patches, loc='upper left', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Cell 5: Visualize Virtual Subtrees for Each Location

fig, axes = plt.subplots(1, 3, figsize=(18, 8))
color_map = {'A': '#e41a1c', 'B': '#377eb8', 'C': '#4daf4a'}

for i, loc in enumerate(['A', 'B', 'C']):
    ax = axes[i]
    
    # Find MRCA for this location
    mrca = find_mrca(root, loc)
    n_tips = root.loc_counts.get(loc, 0)
    
    # Plot tree with this location highlighted
    plot_tree(root, ax, 
              title=f"Virtual Subtree: Location {loc}\n({n_tips} tips, MRCA shown)",
              highlight_loc=loc, show_labels=True, color_map=color_map)
    
    # Mark MRCA
    if mrca:
        y_pos = compute_y_positions(root)
        ax.scatter(mrca.root_distance, y_pos[mrca.id], 
                  marker='s', s=150, c=color_map[loc], 
                  edgecolors='black', linewidths=2, zorder=10,
                  label='MRCA')
        ax.annotate('MRCA', (mrca.root_distance, y_pos[mrca.id]),
                   xytext=(-30, 10), textcoords='offset points',
                   fontsize=9, fontweight='bold',
                   arrowprops=dict(arrowstyle='->', color='black'))

plt.suptitle('Virtual Subtrees by Location (highlighted branches belong to virtual subtree)', 
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Cell 6: Demonstrate CBLV Encoding Process for One Location

target_loc = 'A'
tree_height = max(t.root_distance for t in tips)

# Get CBLV encoding with trace
cblv, trace, mrca = encode_cblv(root, target_loc, tree_height)

print(f"{'='*70}")
print(f"CBLV ENCODING FOR LOCATION {target_loc}")
print(f"{'='*70}")
print(f"Number of tips: {root.loc_counts[target_loc]}")
print(f"MRCA: {mrca.name if mrca else 'None'}")
print(f"Tree height: {tree_height:.2f}")
print(f"\n{'='*70}")
print("TRAVERSAL TRACE (Virtual Inorder with Ladderization)")
print(f"{'='*70}")

leaf_order = []
for item in trace:
    if item[0] == 'leaf':
        _, node, height, accum = item
        leaf_order.append(node.name)
        print(f"  LEAF: {node.name:<12} | root_dist={node.root_distance:.2f} | height_from_last={height:.2f} | accum_edge={accum:.2f}")
    elif item[0] == 'internal':
        _, node, root_dist, accum = item
        print(f"  INTERNAL: {node.name:<9} | root_dist={root_dist:.2f} | accum_edge={accum:.2f}")
    elif item[0] == 'collapse':
        _, node = item
        print(f"  [COLLAPSED]: {node.name} (only 1 relevant child, skipped)")

print(f"\nTraversal order of tips: {' -> '.join(leaf_order)}")
```

```python
# Cell 7: Visualize CBLV Matrix

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
color_map = {'A': '#e41a1c', 'B': '#377eb8', 'C': '#4daf4a'}

for i, loc in enumerate(['A', 'B', 'C']):
    ax = axes[i]
    
    cblv, trace, mrca = encode_cblv(root, loc, tree_height)
    n_tips = root.loc_counts[loc]
    
    # Plot CBLV matrix as heatmap
    im = ax.imshow(cblv, aspect='auto', cmap='Blues', vmin=0, vmax=1)
    
    # Labels
    ax.set_xlabel('CBLV Columns', fontsize=10)
    ax.set_ylabel('Tip Index (inorder)', fontsize=10)
    ax.set_title(f'Location {loc} CBLV\n({n_tips} tips x 4 cols)', 
                fontsize=11, fontweight='bold', color=color_map[loc])
    
    # Column labels
    ax.set_xticks([0, 1, 2, 3])
    ax.set_xticklabels(['leaf_height', 'int_root_dist', 'leaf_edge', 'int_edge'], 
                       fontsize=8, rotation=45, ha='right')
    
    # Row labels (tip names)
    tip_names = []
    for item in trace:
        if item[0] == 'leaf':
            tip_names.append(item[1].name)
    ax.set_yticks(range(len(tip_names)))
    ax.set_yticklabels(tip_names, fontsize=8)
    
    # Add colorbar
    plt.colorbar(im, ax=ax, shrink=0.8)

plt.suptitle('CBLV Matrices for Each Location (normalized by tree height)', 
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Cell 8: Detailed CBLV Table

print(f"{'='*90}")
print("CBLV MATRICES (Detailed View)")
print(f"{'='*90}")

for loc in ['A', 'B', 'C']:
    cblv, trace, mrca = encode_cblv(root, loc, tree_height)
    n_tips = root.loc_counts[loc]
    
    # Get tip names in order
    tip_names = [item[1].name for item in trace if item[0] == 'leaf']
    
    print(f"\nLocation {loc} ({n_tips} tips):")
    print(f"  MRCA: {mrca.name if mrca else 'N/A'}")
    print(f"  {'Row':<4} {'Tip':<12} {'leaf_height':>12} {'int_root_d':>12} {'leaf_edge':>12} {'int_edge':>12}")
    print(f"  {'-'*68}")
    
    for row_idx in range(min(n_tips, len(tip_names))):
        tip_name = tip_names[row_idx] if row_idx < len(tip_names) else '-'
        row = cblv[row_idx]
        print(f"  {row_idx:<4} {tip_name:<12} {row[0]:>12.4f} {row[1]:>12.4f} {row[2]:>12.4f} {row[3]:>12.4f}")
```

```python
# Cell 9: Ladderization Demonstration

print(f"{'='*80}")
print("LADDERIZATION DEMONSTRATION")
print(f"{'='*80}")
print("\nLadderization ensures deterministic encoding by sorting children")
print("at each branching point by loc_max_dist (descending).")
print("\nThis means: child with DEEPEST tip (for this location) is visited FIRST.")

for loc in ['A', 'B', 'C']:
    print(f"\n{'='*60}")
    print(f"Location {loc}:")
    print(f"{'='*60}")
    
    mrca = find_mrca(root, loc)
    if not mrca:
        print("  No MRCA (< 2 tips)")
        continue
    
    # Show branching points and their child ordering
    def show_branching_points(node, depth=0):
        if node.is_leaf:
            return
        
        if is_branching_point(node, loc):
            children = [c for c in node.children if c.loc_counts.get(loc, 0) > 0]
            
            print(f"\n  Branching point: {node.name}")
            print(f"  Children with location {loc} tips:")
            
            for c in children:
                max_dist = c.loc_max_dist.get(loc, 0)
                print(f"    - {c.name}: loc_max_dist[{loc}] = {max_dist:.2f}, tips = {c.loc_counts.get(loc, 0)}")
            
            # Show sorted order
            sorted_children = sorted(children, key=lambda c: c.loc_max_dist.get(loc, 0), reverse=True)
            print(f"  Ladderized order: {' -> '.join(c.name for c in sorted_children)}")
        
        for child in node.children:
            if child.loc_counts.get(loc, 0) > 0:
                show_branching_points(child, depth + 1)
    
    show_branching_points(mrca)
```

```python
# Cell 10: Branch Collapsing Demonstration

print(f"{'='*80}")
print("BRANCH COLLAPSING DEMONSTRATION")
print(f"{'='*80}")
print("\nWhen an internal node has only 1 child with tips from the target location,")
print("it is COLLAPSED (not recorded as a coalescence event).")
print("\nThis ensures the virtual subtree remains strictly bifurcating.")

for loc in ['A', 'B', 'C']:
    print(f"\n{'='*60}")
    print(f"Location {loc}:")
    print(f"{'='*60}")
    
    _, trace, mrca = encode_cblv(root, loc, tree_height)
    
    collapsed = [item for item in trace if item[0] == 'collapse']
    branching = [item for item in trace if item[0] == 'internal']
    leaves = [item for item in trace if item[0] == 'leaf']
    
    print(f"  Tips (leaves): {len(leaves)}")
    print(f"  Branching points (kept): {len(branching)}")
    print(f"  Collapsed nodes (skipped): {len(collapsed)}")
    
    if collapsed:
        print(f"\n  Collapsed nodes:")
        for item in collapsed:
            node = item[1]
            relevant_children = [c for c in node.children if c.loc_counts.get(loc, 0) > 0]
            print(f"    - {node.name}: only 1 relevant child ({relevant_children[0].name if relevant_children else '?'})")
    
    print(f"\n  Verification: {len(leaves)} tips should have {len(leaves)-1} internal nodes")
    print(f"  Actual internal nodes in CBLV: {len(branching)} {'✓' if len(branching) == len(leaves)-1 else '✗'}")
```

```python
# Cell 11: Summary Diagram

fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# Top-left: Original tree
ax = axes[0, 0]
plot_tree(root, ax, title="1. Original Tree", show_labels=False)
legend_patches = [mpatches.Patch(color=color_map[loc], label=f'Loc {loc}') for loc in ['A', 'B', 'C']]
ax.legend(handles=legend_patches, loc='upper left', fontsize=9)

# Top-right: Virtual subtree for A
ax = axes[0, 1]
plot_tree(root, ax, title="2. Virtual Subtree (Location A)", highlight_loc='A', show_labels=True)
ax.text(0.02, 0.02, 'Gray = ignored\nColored = kept\nCollapsed nodes not shown', 
        transform=ax.transAxes, fontsize=8, verticalalignment='bottom',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# Bottom-left: CBLV matrix for A
ax = axes[1, 0]
cblv_a, trace_a, _ = encode_cblv(root, 'A', tree_height)
im = ax.imshow(cblv_a, aspect='auto', cmap='Reds', vmin=0, vmax=1)
ax.set_title("3. CBLV Matrix (Location A)", fontsize=11, fontweight='bold')
ax.set_xlabel('Columns')
ax.set_ylabel('Tip Index')
ax.set_xticks([0, 1, 2, 3])
ax.set_xticklabels(['leaf_h', 'int_rd', 'leaf_e', 'int_e'], fontsize=9)
plt.colorbar(im, ax=ax, shrink=0.8)

# Bottom-right: Summary statistics
ax = axes[1, 1]
ax.axis('off')

summary_text = """
CBLV ENCODING SUMMARY
=====================

For each location X:

1. FIND MRCA
   Deepest node containing ALL tips of location X

2. BUILD VIRTUAL SUBTREE
   - Keep only tips from location X
   - Collapse nodes with 1 relevant child
   - Result: strictly bifurcating tree

3. LADDERIZE
   At each branching point, sort children by
   loc_max_dist[X] (descending)
   -> Deepest tip visited first

4. INORDER TRAVERSAL
   Left subtree -> Node -> Right subtree
   Yields: leaf, internal, leaf, internal, ...

5. ENCODE TO MATRIX
   n_tips rows x 4 columns:
   [leaf_height, int_root_dist, leaf_edge, int_edge]

PROPERTIES:
- Same tree + same location = same CBLV (deterministic)
- n tips -> n rows, n-1 internal nodes
- Values normalized by tree height [0, 1]
"""

ax.text(0.1, 0.95, summary_text, transform=ax.transAxes, fontsize=10,
        verticalalignment='top', fontfamily='monospace',
        bbox=dict(boxstyle='round', facecolor='lightgray', alpha=0.3))

plt.suptitle('Location-Specific CBLV Extraction Process', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Cell 12: Final Comparison - All Locations Side by Side

fig = plt.figure(figsize=(18, 10))

# Create grid: 2 rows, 3 columns
# Top row: virtual subtrees
# Bottom row: CBLV matrices

for i, loc in enumerate(['A', 'B', 'C']):
    # Virtual subtree
    ax1 = fig.add_subplot(2, 3, i + 1)
    plot_tree(root, ax1, title=f"Location {loc} Virtual Subtree", 
              highlight_loc=loc, show_labels=True, color_map=color_map)
    
    # CBLV matrix
    ax2 = fig.add_subplot(2, 3, i + 4)
    cblv, trace, mrca = encode_cblv(root, loc, tree_height)
    
    cmap = {'A': 'Reds', 'B': 'Blues', 'C': 'Greens'}[loc]
    im = ax2.imshow(cblv, aspect='auto', cmap=cmap, vmin=0, vmax=1)
    
    tip_names = [item[1].name for item in trace if item[0] == 'leaf']
    ax2.set_yticks(range(len(tip_names)))
    ax2.set_yticklabels(tip_names, fontsize=7)
    ax2.set_xticks([0, 1, 2, 3])
    ax2.set_xticklabels(['lh', 'ir', 'le', 'ie'], fontsize=8)
    ax2.set_title(f"Location {loc} CBLV ({len(tip_names)} x 4)", fontsize=10, fontweight='bold')
    ax2.set_xlabel('Columns')
    ax2.set_ylabel('Tips (inorder)')
    plt.colorbar(im, ax=ax2, shrink=0.6)

plt.suptitle('Location-Specific CBLV: Virtual Subtrees and Encoded Matrices', 
             fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

print("\nColumn abbreviations: lh=leaf_height, ir=int_root_dist, le=leaf_edge, ie=int_edge")
```
