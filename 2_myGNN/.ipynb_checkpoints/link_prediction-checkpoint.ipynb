# Link Prediction: Migration Rate Prediction

Compare MLP models with and without edge features.

- **Models**: MLP (node only) vs MLP (node + edge features)
- **Features**: Node and edge features (Z-score normalized)
- **Labels**: Migration rates (Z-score normalized)

## Data Loading

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import dgl
import pickle
from sklearn.model_selection import train_test_split

# Load graphs from my_graphs.bin
graph_list, _ = dgl.load_graphs('my_graphs.bin')
with open('my_graphs.pkl', 'rb') as f:
    metadata = pickle.load(f)

all_graphs = {gid: g for gid, g in zip(metadata['graph_ids'], graph_list)}
print(f"Loaded {len(all_graphs)} graphs")

# Filter graphs: keep only those where ALL edges have non-NaN patristic_distance_count
graphs = {}
for gid, g in all_graphs.items():
    if 'patristic_distance_count' in g.edata:
        if not torch.isnan(g.edata['patristic_distance_count']).any():
            graphs[gid] = g

print(f"Filtered to {len(graphs)} valid graphs")
```

## Data Preparation and Normalization

```python
node_features = ['Initial_Population', 'Epidemic_Peak', 'Peak_Timing', 'Accumulated_Infections']
edge_features = ['patristic_distance_min', 'patristic_distance_q25', 'patristic_distance_median', 
                 'patristic_distance_q75', 'patristic_distance_max', 'patristic_distance_mean', 
                 'patristic_distance_std', 'patristic_distance_count']

def prepare_edge_data(graph):
    """Extract src/dst node features, edge features, and labels (filter self-loops)."""
    node_feat = torch.stack([graph.ndata[name] for name in node_features], dim=1)
    edge_feat = torch.stack([graph.edata[name] for name in edge_features], dim=1)
    src, dst = graph.edges()
    labels = graph.edata['migration_rate']
    mask = ~torch.isnan(labels)
    return node_feat[src[mask]], edge_feat[mask], node_feat[dst[mask]], labels[mask]

# Train/val/test split
torch.manual_seed(42)
graph_ids = list(graphs.keys())
train_ids, temp_ids = train_test_split(graph_ids, test_size=0.3, random_state=42)
val_ids, test_ids = train_test_split(temp_ids, test_size=0.5, random_state=42)

train_graphs = [graphs[gid] for gid in train_ids]
val_graphs = [graphs[gid] for gid in val_ids]
test_graphs = [graphs[gid] for gid in test_ids]

# Z-score normalization for node features
print("Node Feature Normalization:")
for fname in node_features:
    train_feat = torch.cat([g.ndata[fname] for g in train_graphs])
    mean, std = train_feat.mean(), train_feat.std()
    for g in train_graphs + val_graphs + test_graphs:
        g.ndata[fname] = (g.ndata[fname] - mean) / std
    print(f"  {fname:30s}: Mean={mean:8.2f}, Std={std:8.2f}")

# Edge feature normalization (with log transform for patristic_distance_count)
print("\nEdge Feature Normalization:")
for fname in edge_features:
    # Apply log transformation to patristic_distance_count
    if fname == 'patristic_distance_count':
        print(f"  {fname}: Applying log(1+x) transformation...")
        for g in train_graphs + val_graphs + test_graphs:
            g.edata[fname] = torch.log1p(g.edata[fname])
    
    # Z-score normalization
    train_feat = torch.cat([g.edata[fname] for g in train_graphs])
    mean, std = train_feat.mean(), train_feat.std()
    for g in train_graphs + val_graphs + test_graphs:
        g.edata[fname] = (g.edata[fname] - mean) / std
    print(f"  {fname:30s}: Mean={mean:8.2f}, Std={std:8.2f}")

# Z-score normalization for labels (migration_rate)
print("\nLabel Normalization:")
train_labels = torch.cat([prepare_edge_data(g)[3] for g in train_graphs])
label_mean, label_std = train_labels.mean(), train_labels.std()

for g in train_graphs + val_graphs + test_graphs:
    mask = ~torch.isnan(g.edata['migration_rate'])
    g.edata['migration_rate'][mask] = (g.edata['migration_rate'][mask] - label_mean) / label_std

print(f"  migration_rate (label)        : Mean={label_mean:.6f}, Std={label_std:.6f}")
print(f"\nSplit: Train={len(train_graphs)}, Val={len(val_graphs)}, Test={len(test_graphs)}")
```

## Model Definitions

```python
class MLP(nn.Module):
    """MLP baseline: [src_node | dst_node] -> prediction."""
    def __init__(self, node_dim):
        super().__init__()
        self.layers = nn.Sequential(
            nn.Linear(node_dim * 2, 32),
            nn.ReLU(),
            nn.Linear(32, 8),
            nn.ReLU(),
            nn.Linear(8, 1)
        )
    
    def forward(self, src_feat, edge_feat, dst_feat):
        x = torch.cat([src_feat, dst_feat], dim=1)
        return self.layers(x).squeeze(-1)


class MLP_EdgeFeat(nn.Module):
    """MLP with edge features: [src_node | edge_feat | dst_node] -> prediction."""
    def __init__(self, node_dim, edge_dim):
        super().__init__()
        self.layers = nn.Sequential(
            nn.Linear(node_dim * 2 + edge_dim, 32),
            nn.ReLU(),
            nn.Linear(32, 8),
            nn.ReLU(),
            nn.Linear(8, 1)
        )
    
    def forward(self, src_feat, edge_feat, dst_feat):
        x = torch.cat([src_feat, edge_feat, dst_feat], dim=1)
        return self.layers(x).squeeze(-1)
```

## Training and Evaluation

```python
from sklearn.metrics import r2_score

def train_epoch(model, graphs, optimizer, criterion):
    model.train()
    total_loss = total_samples = 0
    for g in graphs:
        src_f, edge_f, dst_f, labels = prepare_edge_data(g)
        if len(labels) == 0:
            continue
        optimizer.zero_grad()
        loss = criterion(model(src_f, edge_f, dst_f), labels)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * len(labels)
        total_samples += len(labels)
    return total_loss / total_samples


def evaluate(model, graphs, criterion):
    model.eval()
    total_loss = total_samples = 0
    all_preds, all_labels = [], []
    with torch.no_grad():
        for g in graphs:
            src_f, edge_f, dst_f, labels = prepare_edge_data(g)
            if len(labels) == 0:
                continue
            preds = model(src_f, edge_f, dst_f)
            loss = criterion(preds, labels)
            total_loss += loss.item() * len(labels)
            total_samples += len(labels)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
    return total_loss / total_samples, r2_score(all_labels, all_preds)


def train_model(model, name, epochs=500, patience=20):
    optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
    criterion = nn.MSELoss()
    best_val_loss = float('inf')
    patience_counter = 0
    
    print(f"\n{'='*60}\nTraining {name}\n{'='*60}")
    print(f"Parameters: {sum(p.numel() for p in model.parameters())}")
    
    for epoch in range(epochs):
        train_loss = train_epoch(model, train_graphs, optimizer, criterion)
        val_loss, val_r2 = evaluate(model, val_graphs, criterion)
        
        if (epoch + 1) % 10 == 0:
            print(f"Epoch {epoch+1:3d} | Train: {train_loss:.6f} | Val: {val_loss:.6f} | R²: {val_r2:.4f}")
        
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_state = model.state_dict().copy()
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f"Early stopping at epoch {epoch+1}")
                break
    
    model.load_state_dict(best_state)
    test_loss, test_r2 = evaluate(model, test_graphs, criterion)
    print(f"Best Val: {best_val_loss:.6f} | Test: {test_loss:.6f} | R²: {test_r2:.4f}")
    return test_loss, test_r2


# Train both models
node_dim = len(node_features)
edge_dim = len(edge_features)

mlp_model = MLP(node_dim)
mlp_edge_model = MLP_EdgeFeat(node_dim, edge_dim)

test_loss_mlp, test_r2_mlp = train_model(mlp_model, "MLP (node only)")
test_loss_mlp_edge, test_r2_mlp_edge = train_model(mlp_edge_model, "MLP (node + edge)")

print(f"\n{'='*60}\nSummary\n{'='*60}")
print(f"MLP (node only)   | MSE: {test_loss_mlp:.6f} | R²: {test_r2_mlp:.4f}")
print(f"MLP (node + edge) | MSE: {test_loss_mlp_edge:.6f} | R²: {test_r2_mlp_edge:.4f}")
print("="*60)
```

## Visualization

```python
import matplotlib.pyplot as plt

def get_predictions(model, graphs):
    model.eval()
    preds, labels = [], []
    with torch.no_grad():
        for g in graphs:
            src_f, edge_f, dst_f, label = prepare_edge_data(g)
            if len(label) == 0:
                continue
            preds.extend(model(src_f, edge_f, dst_f).cpu().numpy())
            labels.extend(label.cpu().numpy())
    return np.array(preds), np.array(labels)

# Get predictions (normalized scale)
preds_mlp, labels = get_predictions(mlp_model, test_graphs)
preds_mlp_edge, _ = get_predictions(mlp_edge_model, test_graphs)

# Denormalize for visualization
preds_mlp_orig = preds_mlp * label_std.item() + label_mean.item()
preds_mlp_edge_orig = preds_mlp_edge * label_std.item() + label_mean.item()
labels_orig = labels * label_std.item() + label_mean.item()

# Plot
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# MLP (node only)
axes[0].scatter(labels_orig, preds_mlp_orig, alpha=0.3, s=10, color='blue')
axes[0].plot([labels_orig.min(), labels_orig.max()], [labels_orig.min(), labels_orig.max()], 'r--', lw=2)
axes[0].set_xlabel('True Migration Rate', fontsize=12)
axes[0].set_ylabel('Predicted', fontsize=12)
axes[0].set_title('MLP (node only)', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)
axes[0].text(0.05, 0.95, f'R² = {test_r2_mlp:.4f}\nMSE = {test_loss_mlp:.6f}',
            transform=axes[0].transAxes, fontsize=11, verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# MLP (node + edge)
axes[1].scatter(labels_orig, preds_mlp_edge_orig, alpha=0.3, s=10, color='green')
axes[1].plot([labels_orig.min(), labels_orig.max()], [labels_orig.min(), labels_orig.max()], 'r--', lw=2)
axes[1].set_xlabel('True Migration Rate', fontsize=12)
axes[1].set_ylabel('Predicted', fontsize=12)
axes[1].set_title('MLP (node + edge)', fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3)
axes[1].text(0.05, 0.95, f'R² = {test_r2_mlp_edge:.4f}\nMSE = {test_loss_mlp_edge:.6f}',
            transform=axes[1].transAxes, fontsize=11, verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.show()

print(f"{'='*60}\nModel Comparison (Denormalized)\n{'='*60}")
print(f"{'Model':<20} {'R²':>8} {'MSE (norm)':>12}")
print(f"{'-'*60}")
print(f"{'MLP (node only)':<20} {test_r2_mlp:>8.4f} {test_loss_mlp:>12.6f}")
print(f"{'MLP (node + edge)':<20} {test_r2_mlp_edge:>8.4f} {test_loss_mlp_edge:>12.6f}")
print("="*60)
```
