# Link Prediction with GAT vs MLP

Predicting edge-level label (`migration_rate`) using Graph Attention Network and comparing with MLP baseline.

## 1. Data Loading

```python
import sys
sys.path.append('/Users/lukelyu/Library/CloudStorage/OneDrive-Emory/emory/phyloGNN/utils')

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import dgl
import dgl.function as fn
import matplotlib.pyplot as plt
from copy import deepcopy
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from graph_loading import load_all_graphs

# Load graphs from CSV files
input_folder = '/Users/lukelyu/Desktop/epidata/200_5'
graphs_dict = load_all_graphs(input_folder)
graph_list_all = list(graphs_dict.values())

# Filter graphs where any node has Num_Samples = 0
graph_list_raw = [g for g in graph_list_all if not (g.ndata['Num_Samples'] == 0).any()]

print(f"Loaded {len(graph_list_raw)}/{len(graph_list_all)} graphs")
print(f"Nodes: {graph_list_raw[0].num_nodes()}, Edges: {graph_list_raw[0].num_edges()}")
```

## 2. Feature Normalization

```python
# Feature configuration
node_features = ['Initial_Population', 'Epidemic_Peak', 'Peak_Timing', 
                 'Accumulated_Infections', 'Monophyletic_Groups', 'Num_Samples']
edge_features = ['dtw_distance', 'dtw_lag_mean', 'dtw_lag_std']
edge_label = 'migration_rate'  # Edge prediction target

# Remove self-loops (they have NaN edge features)
graph_list = []
for g in graph_list_raw:
    src, dst = g.edges()
    mask = src != dst
    new_g = dgl.graph((src[mask], dst[mask]), num_nodes=g.num_nodes())
    for key, val in g.ndata.items():
        new_g.ndata[key] = val.clone()
    for key, val in g.edata.items():
        new_g.edata[key] = val[mask].clone()
    graph_list.append(new_g)

# Stack features into tensors
# Scale migration_rate by 1e5 to avoid numerical issues (will scale back for display)
LABEL_SCALE = 1e5
for g in graph_list:
    g.ndata['feat'] = torch.stack([g.ndata[f] for f in node_features], dim=1)
    g.edata['feat'] = torch.stack([g.edata[f] for f in edge_features], dim=1)
    g.edata['label'] = g.edata[edge_label] * LABEL_SCALE  # Scale up

# Train/val/test split (70/15/15)
torch.manual_seed(42)
train_graphs, temp = train_test_split(graph_list, test_size=0.3, random_state=42)
val_graphs, test_graphs = train_test_split(temp, test_size=0.5, random_state=42)

# Store ranges before normalization
all_graphs = train_graphs + val_graphs + test_graphs
node_feats_before = torch.cat([g.ndata['feat'] for g in all_graphs])
edge_feats_before = torch.cat([g.edata['feat'] for g in all_graphs])
node_ranges_before = [(node_feats_before[:, i].min().item(), node_feats_before[:, i].max().item()) 
                      for i in range(len(node_features))]
edge_ranges_before = [(edge_feats_before[:, i].min().item(), edge_feats_before[:, i].max().item()) 
                      for i in range(len(edge_features))]

# Z-score normalization (fit on train, apply to all)
for i in range(len(node_features)):
    train_vals = torch.cat([g.ndata['feat'][:, i] for g in train_graphs])
    mean, std = train_vals.mean(), train_vals.std()
    for g in all_graphs:
        g.ndata['feat'][:, i] = (g.ndata['feat'][:, i] - mean) / std

for i in range(len(edge_features)):
    train_vals = torch.cat([g.edata['feat'][:, i] for g in train_graphs])
    mean, std = train_vals.mean(), train_vals.std()
    for g in all_graphs:
        g.edata['feat'][:, i] = (g.edata['feat'][:, i] - mean) / std

# Print feature ranges
node_feats_after = torch.cat([g.ndata['feat'] for g in all_graphs])
edge_feats_after = torch.cat([g.edata['feat'] for g in all_graphs])
edge_labels_all = torch.cat([g.edata['label'] for g in all_graphs])

print(f"Split: Train={len(train_graphs)}, Val={len(val_graphs)}, Test={len(test_graphs)}")
print(f"Edge label: {edge_label} (scaled by {LABEL_SCALE:.0e})")
print(f"  Scaled range: [{edge_labels_all.min():.2f}, {edge_labels_all.max():.2f}]")
print(f"\nNode features ({len(node_features)}) - before vs after normalization:")
for i, feat in enumerate(node_features):
    before = f"[{node_ranges_before[i][0]:.1f}, {node_ranges_before[i][1]:.1f}]"
    after = f"[{node_feats_after[:, i].min():.2f}, {node_feats_after[:, i].max():.2f}]"
    print(f"  {feat:<25} {before:>22} -> {after}")

print(f"\nEdge features ({len(edge_features)}) - before vs after normalization:")
for i, feat in enumerate(edge_features):
    before = f"[{edge_ranges_before[i][0]:.1f}, {edge_ranges_before[i][1]:.1f}]"
    after = f"[{edge_feats_after[:, i].min():.2f}, {edge_feats_after[:, i].max():.2f}]"
    print(f"  {feat:<25} {before:>22} -> {after}")
```

## 3. Model Definition

Two models:
- **GAT**: 1 round message passing with edge-based attention, concat [self | neighbors], then link prediction
- **MLP**: Baseline, concat raw features [src | edge | dst]

```python
node_dim = len(node_features)
edge_dim = len(edge_features)

class GATLinkPredictor(nn.Module):
    """GAT link predictor with 1 round of message passing (concat aggregation)."""
    def __init__(self, node_in_dim, edge_in_dim, hidden_dim=16):
        super().__init__()
        # Node/edge projection
        self.node_fc = nn.Linear(node_in_dim, hidden_dim)
        self.edge_fc = nn.Linear(edge_in_dim, hidden_dim)
        
        # GAT attention: edge_feat (3) -> 8 -> 1
        self.attn_fc = nn.Sequential(
            nn.Linear(edge_in_dim, 8), nn.ReLU(), nn.Linear(8, 1))
        
        # Classifier: [src_concat | edge | dst_concat] = [32 | 16 | 32] = 80D -> 1
        self.classifier = nn.Sequential(
            nn.ReLU(),
            nn.Linear(hidden_dim * 5, 1)
        )
    
    def forward(self, g, node_feat, edge_feat):
        # 1. Project node features
        h = self.node_fc(node_feat)  # (num_nodes, 16)
        
        # 2. GAT message passing: attention from edge features
        g.ndata['h'] = h
        g.edata['a'] = dgl.ops.edge_softmax(g, self.attn_fc(edge_feat))
        g.update_all(fn.u_mul_e('h', 'a', 'm'), fn.sum('m', 'agg'))
        
        # 3. Concat: [self | neighbors]
        h_updated = torch.cat([h, g.ndata['agg']], dim=1)  # (num_nodes, 32)
        
        # 4. Get src/dst embeddings for each edge
        src, dst = g.edges()
        h_src = h_updated[src]  # (num_edges, 32)
        h_dst = h_updated[dst]  # (num_edges, 32)
        h_edge = self.edge_fc(edge_feat)  # (num_edges, 16)
        
        # 5. Concat and classify: [32 | 16 | 32] = 80D -> 1
        h_concat = torch.cat([h_src, h_edge, h_dst], dim=1)
        return self.classifier(h_concat).squeeze(-1)


class MLPLinkPredictor(nn.Module):
    """MLP baseline: concat raw features [src | edge | dst] -> 1."""
    def __init__(self, node_in_dim, edge_in_dim, hidden_dim=16):
        super().__init__()
        # Match GAT input size: 80D (to be fair)
        self.layers = nn.Sequential(
            nn.Linear(node_in_dim * 2 + edge_in_dim, hidden_dim * 5),
            nn.ReLU(),
            nn.Linear(hidden_dim * 5, 1)
        )
    
    def forward(self, g, node_feat, edge_feat):
        src, dst = g.edges()
        h_concat = torch.cat([node_feat[src], edge_feat, node_feat[dst]], dim=1)
        return self.layers(h_concat).squeeze(-1)


# Initialize models
gat_model = GATLinkPredictor(node_dim, edge_dim)
mlp_model = MLPLinkPredictor(node_dim, edge_dim)

print(f"GAT LinkPredictor: {sum(p.numel() for p in gat_model.parameters())} params")
print(f"MLP LinkPredictor: {sum(p.numel() for p in mlp_model.parameters())} params")

# Check migration rate scale
all_labels = torch.cat([g.edata['label'] for g in all_graphs])
print(f"\nMigration rate statistics (scaled):")
print(f"  Range: [{all_labels.min():.2f}, {all_labels.max():.2f}]")
print(f"  Mean: {all_labels.mean():.2f}, Std: {all_labels.std():.2f}")
```

## 4. Training & Evaluation

```python
def train_model(model, lr=0.001, epochs=500, patience=20, verbose=True):
    """Train model with early stopping. Returns (test_loss, val_loss, model)."""
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    criterion = nn.MSELoss()
    best_val, best_state, wait = float('inf'), None, 0
    
    for epoch in range(epochs):
        # Train
        model.train()
        for g in train_graphs:
            optimizer.zero_grad()
            pred = model(g, g.ndata['feat'], g.edata['feat'])
            loss = criterion(pred, g.edata['label'])
            loss.backward()
            optimizer.step()
        
        # Validate
        model.eval()
        with torch.no_grad():
            val_loss = sum(
                criterion(model(g, g.ndata['feat'], g.edata['feat']),
                         g.edata['label']).item() * g.num_edges() for g in val_graphs
            ) / sum(g.num_edges() for g in val_graphs)
        
        if val_loss < best_val:
            best_val, best_state, wait = val_loss, deepcopy(model.state_dict()), 0
        else:
            wait += 1
            if wait >= patience:
                if verbose: print(f"  Early stopping at epoch {epoch+1}")
                break
        
        if verbose and (epoch + 1) % 50 == 0:
            print(f"  Epoch {epoch+1:3d} | Val: {val_loss:.6f}")
    
    model.load_state_dict(best_state)
    
    # Test
    model.eval()
    with torch.no_grad():
        test_loss = sum(
            criterion(model(g, g.ndata['feat'], g.edata['feat']),
                     g.edata['label']).item() * g.num_edges() for g in test_graphs
        ) / sum(g.num_edges() for g in test_graphs)
    
    if verbose: print(f"  Best Val: {best_val:.6f} | Test: {test_loss:.6f}")
    return test_loss, best_val, model


def get_predictions(model, graphs):
    """Get predictions and labels for evaluation."""
    model.eval()
    preds, labels = [], []
    with torch.no_grad():
        for g in graphs:
            pred = model(g, g.ndata['feat'], g.edata['feat'])
            preds.extend(pred.cpu().numpy())
            labels.extend(g.edata['label'].cpu().numpy())
    return np.array(preds), np.array(labels)


# Train models
print("Training GAT LinkPredictor...")
test_gat, _, gat_model = train_model(gat_model)

print("\nTraining MLP LinkPredictor...")
mlp_model = MLPLinkPredictor(node_dim, edge_dim)  # Re-initialize
test_mlp, _, mlp_model = train_model(mlp_model)

print(f"\n{'='*40}")
print(f"GAT Test MSE: {test_gat:.6f}")
print(f"MLP Test MSE: {test_mlp:.6f}")
print(f"{'='*40}")
```

## 5. Visualization

```python
# Plot results (scale back to original migration rate values)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

models = [('GAT', gat_model, test_gat, 'orange'), 
          ('MLP', mlp_model, test_mlp, 'green')]

for ax, (name, model, loss, color) in zip(axes, models):
    preds, labels = get_predictions(model, test_graphs)
    # Scale back to original values for display
    preds_orig = preds / LABEL_SCALE
    labels_orig = labels / LABEL_SCALE
    r2 = r2_score(labels, preds)  # R² is scale-invariant
    
    ax.scatter(labels_orig, preds_orig, alpha=0.3, color=color, s=5)
    ax.plot([labels_orig.min(), labels_orig.max()], [labels_orig.min(), labels_orig.max()], 'r--', lw=2)
    ax.set_xlabel('True Migration Rate')
    ax.set_ylabel('Predicted Migration Rate')
    ax.set_title(f'{name}')
    ax.grid(True, alpha=0.3)
    ax.text(0.05, 0.95, f'R²={r2:.4f}\nMSE={loss:.4f} (scaled)', transform=ax.transAxes,
            va='top', bbox=dict(boxstyle='round', fc='wheat', alpha=0.5))
    # Scientific notation for axes
    ax.ticklabel_format(style='scientific', axis='both', scilimits=(0,0))

plt.tight_layout()
plt.show()

# Summary table
print(f"\n{'Model':<8} {'Params':>8} {'Test MSE (scaled)':>18} {'R²':>8}")
print("-"*46)
for name, model, loss, _ in models:
    preds, labels = get_predictions(model, test_graphs)
    params = sum(p.numel() for p in model.parameters())
    print(f"{name:<8} {params:>8} {loss:>18.4f} {r2_score(labels, preds):>8.4f}")
```
