# Link Prediction: Migration Rate Prediction

Predict migration rates between nodes using node features only.

- **Task**: Edge-level regression (link prediction)
- **Node features**: 4 features with Z-score normalization
- **Edge labels**: Migration rates (normalized to prevent scale issues)
- **Model**: MLP on concatenated source/destination node features

## Data Loading and Filtering

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import dgl
import pickle
from pathlib import Path
from sklearn.model_selection import train_test_split

def load_graphs(input_path):
    """Load saved DGL graphs and filter valid ones."""
    input_path = Path(input_path)
    
    # Load graphs
    graph_list, _ = dgl.load_graphs(str(input_path))
    
    # Load metadata
    with open(input_path.with_suffix('.pkl'), 'rb') as f:
        metadata = pickle.load(f)
    
    graph_ids = metadata['graph_ids']
    all_graphs = {gid: g for gid, g in zip(graph_ids, graph_list)}
    
    # Filter: keep only graphs with valid patristic_distance_count
    graphs = {}
    for gid, g in all_graphs.items():
        if 'patristic_distance_count' in g.edata:
            if not torch.isnan(g.edata['patristic_distance_count']).any():
                graphs[gid] = g
    
    print(f"Loaded {len(all_graphs)} graphs, filtered to {len(graphs)} valid graphs")
    return graphs

# Load and filter data
graphs = load_graphs('my_graphs.bin')

# Check label distribution
all_labels = []
for g in graphs.values():
    labels = g.edata['migration_rate']
    valid_labels = labels[~torch.isnan(labels)]
    all_labels.extend(valid_labels.tolist())

all_labels = np.array(all_labels)
print(f"\nMigration rate statistics:")
print(f"  Min: {all_labels.min():.6f}")
print(f"  Max: {all_labels.max():.6f}")
print(f"  Mean: {all_labels.mean():.6f}")
print(f"  Std: {all_labels.std():.6f}")
print(f"  Range: {all_labels.max() - all_labels.min():.6f}")
```

## Feature Preparation and Normalization

```python
# Define features
node_features = ['Initial_Population', 'Epidemic_Peak', 'Peak_Timing', 'Accumulated_Infections']

# Train/val/test split
torch.manual_seed(42)
graph_ids = list(graphs.keys())
train_ids, temp_ids = train_test_split(graph_ids, test_size=0.3, random_state=42)
val_ids, test_ids = train_test_split(temp_ids, test_size=0.5, random_state=42)

train_graphs = [graphs[gid] for gid in train_ids]
val_graphs = [graphs[gid] for gid in val_ids]
test_graphs = [graphs[gid] for gid in test_ids]

# Z-score normalization for node features
print(f"\n{'='*60}\nNode Feature Normalization\n{'='*60}")
for fname in node_features:
    train_feat = torch.cat([g.ndata[fname] for g in train_graphs])
    mean, std = train_feat.mean(), train_feat.std()
    
    for g in train_graphs + val_graphs + test_graphs:
        g.ndata[fname] = (g.ndata[fname] - mean) / std
    print(f"{fname:30s}: Mean={mean:8.2f}, Std={std:8.2f}")

# CRITICAL: Normalize migration_rate labels
print(f"\n{'='*60}\nEdge Label Normalization (migration_rate)\n{'='*60}")
train_labels = []
for g in train_graphs:
    labels = g.edata['migration_rate']
    valid_labels = labels[~torch.isnan(labels)]
    train_labels.extend(valid_labels.tolist())

train_labels = torch.tensor(train_labels)
label_mean = train_labels.mean()
label_std = train_labels.std()

print(f"Raw labels - Mean: {label_mean:.6f}, Std: {label_std:.6f}")

# Normalize labels
for g in train_graphs + val_graphs + test_graphs:
    mask = ~torch.isnan(g.edata['migration_rate'])
    g.edata['migration_rate'][mask] = (g.edata['migration_rate'][mask] - label_mean) / label_std

print(f"Labels normalized to zero mean, unit variance")
print(f"\nSplit: Train={len(train_graphs)}, Val={len(val_graphs)}, Test={len(test_graphs)}")
```

## Model Definition

```python
class LinkPredictorMLP(nn.Module):
    """MLP for link prediction: [src_feat | dst_feat] -> migration_rate."""
    def __init__(self, in_dim, hidden_dim=32):
        super().__init__()
        self.predictor = nn.Sequential(
            nn.Linear(in_dim * 2, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, 8),
            nn.ReLU(),
            nn.Linear(8, 1)
        )
    
    def forward(self, src_feat, dst_feat):
        x = torch.cat([src_feat, dst_feat], dim=1)
        return self.predictor(x).squeeze(-1)


def prepare_edge_data(graph, node_features):
    """Extract src/dst features and labels for edges (excluding self-loops)."""
    # Stack node features
    node_feat = torch.stack([graph.ndata[name] for name in node_features], dim=1)
    
    # Get edges and labels
    src, dst = graph.edges()
    labels = graph.edata['migration_rate']
    
    # Filter self-loops (NaN labels)
    mask = ~torch.isnan(labels)
    src_feat = node_feat[src[mask]]
    dst_feat = node_feat[dst[mask]]
    labels = labels[mask]
    
    return src_feat, dst_feat, labels
```

## Training and Evaluation

```python
from sklearn.metrics import r2_score

def train_epoch(model, graphs, optimizer, criterion):
    model.train()
    total_loss = total_samples = 0
    
    for g in graphs:
        src_f, dst_f, labels = prepare_edge_data(g, node_features)
        if len(labels) == 0:
            continue
        
        optimizer.zero_grad()
        loss = criterion(model(src_f, dst_f), labels)
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item() * len(labels)
        total_samples += len(labels)
    
    return total_loss / total_samples


def evaluate(model, graphs, criterion):
    """Evaluate model and return MSE and R²."""
    model.eval()
    total_loss = total_samples = 0
    all_preds, all_labels = [], []
    
    with torch.no_grad():
        for g in graphs:
            src_f, dst_f, labels = prepare_edge_data(g, node_features)
            if len(labels) == 0:
                continue
            
            preds = model(src_f, dst_f)
            loss = criterion(preds, labels)
            
            total_loss += loss.item() * len(labels)
            total_samples += len(labels)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
    
    mse = total_loss / total_samples
    r2 = r2_score(all_labels, all_preds)
    
    return mse, r2


def train_model(model, name, epochs=300, patience=20):
    optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
    criterion = nn.MSELoss()
    best_val_loss = float('inf')
    patience_counter = 0
    
    print(f"\n{'='*60}\nTraining {name}\n{'='*60}")
    print(f"Parameters: {sum(p.numel() for p in model.parameters())}")
    
    for epoch in range(epochs):
        train_loss = train_epoch(model, train_graphs, optimizer, criterion)
        val_loss, val_r2 = evaluate(model, val_graphs, criterion)
        
        if (epoch + 1) % 10 == 0:
            print(f"Epoch {epoch+1:3d} | Train: {train_loss:.6f} | Val: {val_loss:.6f} | R²: {val_r2:.4f}")
        
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_state = model.state_dict().copy()
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f"Early stopping at epoch {epoch+1}")
                break
    
    model.load_state_dict(best_state)
    test_loss, test_r2 = evaluate(model, test_graphs, criterion)
    print(f"Best Val: {best_val_loss:.6f} | Test: {test_loss:.6f} | R²: {test_r2:.4f}")
    
    return test_loss, test_r2


# Train model
in_dim = len(node_features)
model = LinkPredictorMLP(in_dim, hidden_dim=32)

test_loss, test_r2 = train_model(model, "Link Predictor MLP")

print(f"\n{'='*60}\nFinal Results\n{'='*60}")
print(f"Test MSE: {test_loss:.6f}")
print(f"Test R²:  {test_r2:.4f}")
print("="*60)
```

## Visualization

```python
import matplotlib.pyplot as plt

def get_predictions(model, graphs):
    model.eval()
    preds, labels = [], []
    with torch.no_grad():
        for g in graphs:
            src_f, dst_f, label = prepare_edge_data(g, node_features)
            if len(label) == 0:
                continue
            preds.extend(model(src_f, dst_f).cpu().numpy())
            labels.extend(label.cpu().numpy())
    return np.array(preds), np.array(labels)

# Get predictions (normalized scale)
preds, labels = get_predictions(model, test_graphs)

# Denormalize for visualization
preds_denorm = preds * label_std.item() + label_mean.item()
labels_denorm = labels * label_std.item() + label_mean.item()

# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Scatter plot (denormalized)
axes[0].scatter(labels_denorm, preds_denorm, alpha=0.3, s=10)
axes[0].plot([labels_denorm.min(), labels_denorm.max()], 
             [labels_denorm.min(), labels_denorm.max()], 'r--', lw=2)
axes[0].set_xlabel('True Migration Rate', fontsize=12)
axes[0].set_ylabel('Predicted Migration Rate', fontsize=12)
axes[0].set_title('Predictions vs True Values', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)

mae_denorm = np.abs(preds_denorm - labels_denorm).mean()
axes[0].text(0.05, 0.95, f'R² = {test_r2:.4f}\nMAE = {mae_denorm:.6f}\nMSE = {test_loss:.6f}',
            transform=axes[0].transAxes, fontsize=11, verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# Residual plot (normalized)
residuals = preds - labels
axes[1].scatter(labels, residuals, alpha=0.3, s=10)
axes[1].axhline(y=0, color='r', linestyle='--', lw=2)
axes[1].set_xlabel('True Migration Rate (normalized)', fontsize=12)
axes[1].set_ylabel('Residuals', fontsize=12)
axes[1].set_title('Residual Plot', fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"{'='*60}\nTest Set Results\n{'='*60}")
print(f"Number of edges: {len(preds):,}")
print(f"R²: {test_r2:.4f}")
print(f"MAE (denormalized): {mae_denorm:.6f}")
print(f"MSE (normalized): {test_loss:.6f}")
print(f"Label range: [{labels_denorm.min():.6f}, {labels_denorm.max():.6f}]")
print("="*60)
```
